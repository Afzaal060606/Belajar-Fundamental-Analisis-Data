## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

Berdasarkan eksplorasi awal terhadap rentang waktu transaksi (September 2016 – Oktober 2018), data bulanan yang utuh dan representatif tercatat selama **20 bulan penuh dari Januari 2017 hingga Agustus 2018** (99,65% dari total pesanan). Oleh karena itu, seluruh pertanyaan analisis dirumuskan dengan kerangka **SMART** pada periode tersebut:

- **Pertanyaan 1 (Performa Kategori Produk & Tren Pendapatan):** Kategori produk apa saja yang menghasilkan total pendapatan (*revenue*) tertinggi dan terendah di platform e-commerce sepanjang periode Januari 2017 hingga Agustus 2018, serta bagaimana tren pertumbuhan pendapatan bulanannya (*Month-over-Month*) untuk menentukan prioritas alokasi anggaran promosi?
  - *Specific:* Fokus pada pendapatan per kategori produk dan laju pertumbuhan bulanan.
  - *Measurable:* Diukur menggunakan total nilai penjualan (`price` dalam BRL) dan persentase perubahan MoM.
  - *Action-Oriented:* Menentukan kategori unggulan untuk kampanye promosi utama serta kategori berkinerja rendah untuk evaluasi katalog atau *bundling*.
  - *Relevant:* Berkaitan langsung dengan optimalisasi pendapatan platform.
  - *Time-bound:* Periode Januari 2017 – Agustus 2018.

- **Pertanyaan 2 (Ketepatan Waktu Pengiriman & Kepuasan Pelanggan):** Seberapa besar perbedaan skor rata-rata kepuasan pelanggan (`review_score`) antara pesanan yang tiba tepat waktu dibandingkan dengan pesanan yang mengalami keterlambatan pengiriman (*late delivery*), serta negara bagian (*state*) mana yang memiliki tingkat keterlambatan tertinggi sepanjang periode Januari 2017 hingga Agustus 2018?
  - *Specific:* Membandingkan `review_score` berdasarkan status ketepatan pengiriman dan memetakan tingkat keterlambatan per wilayah negara bagian.
  - *Measurable:* Diukur melalui selisih rata-rata skor ulasan (skala 1–5) dan persentase pesanan terlambat (%).
  - *Action-Oriented:* Menjadi dasar perbaikan SLA logistik dan pemilihan mitra kurir di wilayah dengan keterlambatan tinggi.
  - *Relevant:* Kecepatan pengiriman merupakan faktor kunci kepuasan dan retensi pelanggan e-commerce.
  - *Time-bound:* Periode Januari 2017 – Agustus 2018.

- **Pertanyaan 3 (Metode Pembayaran & Nilai Transaksi):** Bagaimana distribusi penggunaan metode pembayaran (`payment_type`) dan kontribusinya terhadap total nilai transaksi (`payment_value`) serta rata-rata nilai pesanan (*Average Order Value*) selama periode Januari 2017 hingga Agustus 2018 untuk merancang program kemitraan finansial dan promo cicilan?
  - *Specific:* Fokus pada pangsa penggunaan dan nilai transaksi tiap metode pembayaran.
  - *Measurable:* Diukur dari persentase jumlah transaksi (%), total nilai pembayaran (BRL), dan rata-rata nilai transaksi (AOV).
  - *Action-Oriented:* Mengarahkan kerja sama promo cicilan kartu kredit atau insentif pembayaran tertentu.
  - *Relevant:* Mendukung peningkatan konversi *checkout* dan nilai keranjang belanja.
  - *Time-bound:* Periode Januari 2017 – Agustus 2018.

### Cleaning Data

#### Fixing ... problem

```python
# 1. Konversi kolom tanggal ke format datetime
order_date_cols = [
    "order_purchase_timestamp", "order_approved_at",
    "order_delivered_carrier_date", "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
for col in order_date_cols:
    orders_df[col] = pd.to_datetime(orders_df[col])

order_items_df["shipping_limit_date"] = pd.to_datetime(order_items_df["shipping_limit_date"])
order_reviews_df["review_creation_date"] = pd.to_datetime(order_reviews_df["review_creation_date"])
order_reviews_df["review_answer_timestamp"] = pd.to_datetime(order_reviews_df["review_answer_timestamp"])

# 2. Melengkapi terjemahan kategori & imputasi missing values products_df
extra_translations = pd.DataFrame({
    "product_category_name": ["pc_gamer", "portateis_cozinha_e_preparadores_de_alimentos"],
    "product_category_name_english": ["pc_gamer", "portable_kitchen_food_preparers"]
})
category_trans_clean = pd.concat([category_trans_df, extra_translations], ignore_index=True).drop_duplicates("product_category_name")

products_clean = products_df.merge(category_trans_clean, on="product_category_name", how="left")
products_clean["product_category_name"] = products_clean["product_category_name"].fillna("unknown")
products_clean["product_category_name_english"] = products_clean["product_category_name_english"].fillna("unknown")

num_prod_cols = [
    "product_name_lenght", "product_description_lenght", "product_photos_qty",
    "product_weight_g", "product_length_cm", "product_height_cm", "product_width_cm"
]
for col in num_prod_cols:
    products_clean[col] = products_clean[col].fillna(products_clean[col].median())

# 3. Imputasi missing values ulasan teks
reviews_clean = order_reviews_df.copy()
reviews_clean["review_comment_title"] = reviews_clean["review_comment_title"].fillna("No Title")
reviews_clean["review_comment_message"] = reviews_clean["review_comment_message"].fillna("No Comment")

# 4. Filter transaksi pembayaran bernilai > 0
payments_clean = order_payments_df[order_payments_df["payment_value"] > 0].copy()

# 5. Deduplikasi & agregasi koordinat geolocation per kode pos
geo_clean = geolocation_df.drop_duplicates().groupby("geolocation_zip_code_prefix", as_index=False).agg(
    lat=("geolocation_lat", "mean"),
    lon=("geolocation_lng", "mean"),
    geolocation_city=("geolocation_city", "first"),
    geolocation_state=("geolocation_state", "first")
)

print("=== VERIFIKASI HASIL CLEANING DATA ===")
print("- Tipe `order_purchase_timestamp`:", orders_df["order_purchase_timestamp"].dtype)
print("- Total missing `products_clean`   :", products_clean.isnull().sum().sum())
print("- Total missing `reviews_clean`    :", reviews_clean.isnull().sum().sum())
print("- Baris unik `geo_clean`           :", f"{len(geo_clean):,}")
print("- Baris valid `payments_clean`     :", f"{len(payments_clean):,}")
```

**Insight:**
- Seluruh kolom tanggal pada tabel pesanan, item, dan ulasan kini bertipe `datetime64[ns]`, sehingga selisih hari pengiriman dapat dihitung secara presisi.
- Tabel `products_clean` dan `reviews_clean` sudah bebas dari *missing value* pada atribut analisis utama, sementara `geo_clean` telah terkompresi menjadi 19.015 titik koordinat kode pos unik yang siap digabungkan ke data pelanggan.

## Exploratory Data Analysis (EDA)

### Explore Performa Kategori Produk & Tren Pendapatan Bulanan

```python
# Filter periode analisis utama: Januari 2017 - Agustus 2018 (20 bulan penuh)
orders_period = orders_df[
    (orders_df["order_purchase_timestamp"] >= "2017-01-01") &
    (orders_df["order_purchase_timestamp"] < "2018-09-01")
].copy()

# Filter pesanan valid (tidak dibatalkan / unavailable) untuk analisis penjualan & pembayaran
orders_valid = orders_period[~orders_period["order_status"].isin(["canceled", "unavailable"])].copy()

# Gabungkan data pesanan valid dengan detail item dan nama kategori bahasa Inggris
sales_df = (
    orders_valid.merge(order_items_df, on="order_id", how="inner")
    .merge(products_clean[["product_id", "product_category_name_english"]], on="product_id", how="left")
)

print("=== STATISTIK DESKRIPTIF HARGA PRODUK & ONGKOS KIRIM (BRL) ===")
display(sales_df[["price", "freight_value"]].describe().round(2))

# Agregasi performa per kategori produk
category_summary = sales_df.groupby("product_category_name_english").agg(
    total_revenue=("price", "sum"),
    total_freight=("freight_value", "sum"),
    total_orders=("order_id", "nunique"),
    total_items=("order_item_id", "count"),
    avg_price=("price", "mean")
).sort_values("total_revenue", ascending=False)

print("\n=== TOP 10 KATEGORI PRODUK BERDASARKAN TOTAL PENDAPATAN ===")
display(category_summary.head(10).round(2))

print("\n=== BOTTOM 10 KATEGORI PRODUK BERDASARKAN TOTAL PENDAPATAN (NON-UNKNOWN) ===")
display(category_summary[category_summary.index != "unknown"].tail(10).round(2))

# Agregasi tren pendapatan bulanan (Month-over-Month)
sales_df["year_month"] = sales_df["order_purchase_timestamp"].dt.to_period("M")
monthly_trend = sales_df.groupby("year_month").agg(
    total_revenue=("price", "sum"),
    total_orders=("order_id", "nunique")
)
monthly_trend["mom_growth_pct"] = monthly_trend["total_revenue"].pct_change() * 100

print("\n=== TREN PENDAPATAN & PERTUMBUHAN BULANAN (JAN 2017 - AGU 2018) ===")
display(monthly_trend.round(2))
```

### Explore Ketepatan Waktu Pengiriman & Skor Kepuasan Pelanggan

```python
# Ambil pesanan berstatus 'delivered' dengan tanggal sampai yang valid
delivered_df = orders_period[orders_period["order_status"] == "delivered"].dropna(
    subset=["order_delivered_customer_date"]
).copy()

# Hitung durasi aktual pengiriman & selisih keterlambatan (hari)
delivered_df["actual_delivery_days"] = (
    delivered_df["order_delivered_customer_date"] - delivered_df["order_purchase_timestamp"]
).dt.total_seconds() / 86400.0
delivered_df["delay_days"] = (
    delivered_df["order_delivered_customer_date"] - delivered_df["order_estimated_delivery_date"]
).dt.total_seconds() / 86400.0
delivered_df["is_late"] = delivered_df["delay_days"] > 0

def classify_delay(d):
    if d <= 0:
        return "1. Tepat Waktu / Lebih Awal"
    elif d <= 3:
        return "2. Terlambat 1-3 Hari"
    elif d <= 7:
        return "3. Terlambat 4-7 Hari"
    return "4. Terlambat >7 Hari"

delivered_df["delay_category"] = delivered_df["delay_days"].apply(classify_delay)

# Gabungkan dengan skor ulasan pelanggan
deliv_review_df = delivered_df.merge(reviews_clean[["order_id", "review_score"]], on="order_id", how="inner")

print("=== PERBANDINGAN SKOR ULASAN: TEPAT WAKTU (False) VS TERLAMBAT (True) ===")
display(deliv_review_df.groupby("is_late")["review_score"].agg(["count", "mean", "median", "std"]).round(2))

print("\n=== RATA-RATA SKOR ULASAN BERDASARKAN TINGKAT KETERLAMBATAN ===")
display(deliv_review_df.groupby("delay_category")["review_score"].agg(["count", "mean", "median"]).round(2))

# Agregasi tingkat keterlambatan per negara bagian (customer_state)
deliv_state_df = delivered_df.merge(customers_df[["customer_id", "customer_state"]], on="customer_id", how="left")
state_sla = deliv_state_df.groupby("customer_state").agg(
    total_delivered=("order_id", "count"),
    late_orders=("is_late", "sum"),
    late_rate_pct=("is_late", lambda x: x.mean() * 100),
    avg_delivery_days=("actual_delivery_days", "mean")
).sort_values("late_rate_pct", ascending=False)

print("\n=== TOP 10 NEGARA BAGIAN DENGAN TINGKAT KETERLAMBATAN TERTINGGI (MIN. 500 PESANAN) ===")
display(state_sla[state_sla["total_delivered"] >= 500].head(10).round(2))
```

### Explore Metode Pembayaran & Nilai Transaksi

```python
pay_valid_df = orders_valid.merge(payments_clean, on="order_id", how="inner")

payment_summary = pay_valid_df.groupby("payment_type").agg(
    transaction_count=("order_id", "count"),
    unique_orders=("order_id", "nunique"),
    total_payment_value=("payment_value", "sum"),
    avg_payment_value=("payment_value", "mean"),
    avg_installments=("payment_installments", "mean")
).sort_values("total_payment_value", ascending=False)

payment_summary["value_share_pct"] = (
    payment_summary["total_payment_value"] / payment_summary["total_payment_value"].sum() * 100
)
payment_summary["count_share_pct"] = (
    payment_summary["transaction_count"] / payment_summary["transaction_count"].sum() * 100
)

print("=== RINGKASAN PERFORMA METODE PEMBAYARAN (JAN 2017 - AGU 2018) ===")
display(payment_summary.round(2))
```

**Insight:**
- **Kategori Produk & Tren Pendapatan:** Kategori `health_beauty` menghasilkan pendapatan tertinggi sebesar **R$ 1.251.145,54** (8.757 pesanan), diikuti `watches_gifts` (**R$ 1.194.824,97**) dan `bed_bath_table` (**R$ 1.035.485,07**, volume tertinggi 9.394 pesanan). Sebaliknya, `security_and_services` (R$ 283,29) dan `fashion_childrens_clothes` (R$ 569,85) mencatat pendapatan terendah. Secara bulanan, pendapatan tumbuh konsisten hingga mencapai puncak pada **November 2017 (R$ 1.003.862,14, naik +52,06% MoM)** saat momen *Black Friday*, lalu stabil di kisaran R$ 837 ribu – R$ 993 ribu per bulan pada Januari–Agustus 2018.
- **Ketepatan Pengiriman & Kepuasan Pelanggan:** Sebanyak **92,01%** pesanan tiba tepat waktu dengan rata-rata `review_score` **4,29**, sedangkan **7,99%** pesanan terlambat anjlok ke rata-rata **2,57** (turun **1,73 poin** atau -40,2%). Keterlambatan >7 hari menekan skor ulasan hingga **1,73**. Di antara negara bagian dengan ≥500 pesanan, tingkat keterlambatan tertinggi terjadi di **Maranhão / MA (19,64%)**, **Ceará / CE (15,40%)**, dan **Bahia / BA (14,05%)**.
- **Metode Pembayaran:** `credit_card` mendominasi **78,46%** total nilai transaksi (**R$ 12,31 juta**) dengan rata-rata nilai transaksi tertinggi (**R$ 162,65**) dan rata-rata cicilan **3,50 kali**, diikuti `boleto` (**17,96%** nilai transaksi; AOV **R$ 144,65**).

## Analisis Lanjutan: RFM Analysis & Geospatial Analysis

Untuk memperdalam pemahaman perilaku pelanggan serta distribusi pasar lintas wilayah tanpa menggunakan algoritma *machine learning*, diterapkan dua teknik analisis lanjutan:

1. **RFM Analysis (Manual Grouping & Binning):** Mengelompokkan **94.703 pelanggan unik** (`customer_unique_id`) pada periode Januari 2017 – Agustus 2018 berdasarkan tiga metrik utama:
   - **Recency (R):** Jumlah hari sejak transaksi terakhir pelanggan hingga tanggal acuan (*snapshot date* = 1 hari setelah transaksi terakhir di dataset).
   - **Frequency (F):** Jumlah total pesanan unik yang dilakukan pelanggan.
   - **Monetary (M):** Total nilai pengeluaran produk (`price` dalam BRL) pelanggan.
   Ketiga metrik diberi skor 1–5 menggunakan teknik *quantile binning* (`pd.qcut`) untuk R dan M serta aturan pengelompokan bisnis (*manual binning*) untuk F, lalu dikelompokkan ke dalam 7 segmen pelanggan.
2. **Geospatial Analysis:** Memetakan konsentrasi volume pesanan, jumlah pelanggan unik, dan kontribusi pendapatan berdasarkan lokasi geografis negara bagian (*state*) serta koordinat lintang/bujur (`lat`, `lon`) di Brasil.

```python
# === 1. RFM ANALYSIS (TANPA MACHINE LEARNING) ===
order_rev_df = order_items_df.groupby("order_id", as_index=False)["price"].sum().rename(columns={"price": "order_revenue"})

rfm_base = (
    orders_valid.merge(customers_df[["customer_id", "customer_unique_id", "customer_state"]], on="customer_id", how="inner")
    .merge(order_rev_df, on="order_id", how="inner")
)

snapshot_date = rfm_base["order_purchase_timestamp"].max() + pd.Timedelta(days=1)
print("Snapshot Date RFM:", snapshot_date)

rfm_df = rfm_base.groupby("customer_unique_id", as_index=False).agg(
    recency=("order_purchase_timestamp", lambda x: (snapshot_date - x.max()).days),
    frequency=("order_id", "nunique"),
    monetary=("order_revenue", "sum")
)

# Pemberian skor R, F, M (skala 1-5)
rfm_df["r_score"] = pd.qcut(rfm_df["recency"].rank(method="first"), q=5, labels=[5, 4, 3, 2, 1]).astype(int)
rfm_df["f_score"] = rfm_df["frequency"].apply(lambda f: 1 if f == 1 else (3 if f == 2 else (4 if f == 3 else 5)))
rfm_df["m_score"] = pd.qcut(rfm_df["monetary"].rank(method="first"), q=5, labels=[1, 2, 3, 4, 5]).astype(int)

# Manual Customer Segmentation berdasarkan aturan bisnis skor RFM
def assign_rfm_segment(row):
    if row["r_score"] >= 4 and row["f_score"] >= 3 and row["m_score"] >= 4:
        return "Champions"
    elif row["f_score"] >= 3:
        return "Loyal Customers"
    elif row["r_score"] >= 4 and row["m_score"] >= 4:
        return "Big Spenders (Recent)"
    elif row["r_score"] >= 4:
        return "Recent Customers"
    elif row["r_score"] == 3:
        return "Promising / Need Attention"
    elif row["r_score"] == 2:
        return "At Risk"
    return "Hibernating / Lost"

rfm_df["segment"] = rfm_df.apply(assign_rfm_segment, axis=1)

rfm_seg_summary = rfm_df.groupby("segment").agg(
    customer_count=("customer_unique_id", "count"),
    avg_recency_days=("recency", "mean"),
    avg_frequency=("frequency", "mean"),
    avg_monetary=("monetary", "mean"),
    total_monetary=("monetary", "sum")
).sort_values("total_monetary", ascending=False)

rfm_seg_summary["customer_share_pct"] = rfm_seg_summary["customer_count"] / rfm_seg_summary["customer_count"].sum() * 100
rfm_seg_summary["revenue_share_pct"] = rfm_seg_summary["total_monetary"] / rfm_seg_summary["total_monetary"].sum() * 100

print("\n=== RINGKASAN SEGMENTASI PELANGGAN (RFM ANALYSIS) ===")
display(rfm_seg_summary.round(2))

# === 2. GEOSPATIAL ANALYSIS (DISTRIBUSI WILAYAH NEGARA BAGIAN) ===
geo_state_summary = rfm_base.groupby("customer_state", as_index=False).agg(
    total_orders=("order_id", "nunique"),
    unique_customers=("customer_unique_id", "nunique"),
    total_revenue=("order_revenue", "sum")
).sort_values("total_revenue", ascending=False)

geo_state_summary["revenue_share_pct"] = geo_state_summary["total_revenue"] / geo_state_summary["total_revenue"].sum() * 100

print("\n=== TOP 10 NEGARA BAGIAN BERDASARKAN TOTAL PENDAPATAN ===")
display(geo_state_summary.head(10).round(2))

# === 3. VISUALISASI RFM & GEOSPATIAL ===
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

seg_plot = rfm_seg_summary.sort_values("customer_count", ascending=True)
seg_colors = ["#a6bddb"] * (len(seg_plot) - 1) + ["#1f77b4"]
axes[0].barh(seg_plot.index, seg_plot["customer_count"], color=seg_colors)
axes[0].set_title("Distribusi Segmen Pelanggan Berdasarkan RFM Analysis\n(Total: 94.703 Pelanggan Unik)", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Jumlah Pelanggan Unik")
axes[0].set_xlim(0, seg_plot["customer_count"].max() * 1.18)
for idx, val in enumerate(seg_plot["customer_count"]):
    axes[0].text(val + 350, idx, f"{val:,}", va="center", fontsize=9)

geo_plot = geo_state_summary.head(10).sort_values("total_revenue", ascending=True)
geo_colors = ["#a6bddb"] * (len(geo_plot) - 1) + ["#1f77b4"]
axes[1].barh(geo_plot["customer_state"], geo_plot["total_revenue"] / 1e6, color=geo_colors)
axes[1].set_title("Top 10 Negara Bagian Berdasarkan Total Pendapatan\n(Geospatial Revenue Distribution)", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Total Pendapatan (Juta BRL)")
axes[1].set_xlim(0, (geo_plot["total_revenue"].max() / 1e6) * 1.18)
for idx, val in enumerate(geo_plot["total_revenue"] / 1e6):
    axes[1].text(val + 0.08, idx, f"R$ {val:.2f}M", va="center", fontsize=9)

sns.despine()
plt.tight_layout()
plt.show()

# === 4. EKSPOR DATASET BERSIH UNTUK DASHBOARD STREAMLIT ===
os.makedirs("dashboard", exist_ok=True)

pay_order_agg = payments_clean.groupby("order_id", as_index=False).agg(
    total_payment=("payment_value", "sum"),
    primary_payment_type=("payment_type", "first"),
    max_installments=("payment_installments", "max")
)
rev_order_agg = reviews_clean.groupby("order_id", as_index=False)["review_score"].mean()

main_data = (
    orders_valid.merge(order_items_df, on="order_id", how="inner")
    .merge(products_clean[["product_id", "product_category_name_english"]], on="product_id", how="left")
    .merge(customers_df[["customer_id", "customer_unique_id", "customer_zip_code_prefix", "customer_city", "customer_state"]], on="customer_id", how="left")
    .merge(pay_order_agg, on="order_id", how="left")
    .merge(rev_order_agg, on="order_id", how="left")
    .merge(geo_clean[["geolocation_zip_code_prefix", "lat", "lon"]], left_on="customer_zip_code_prefix", right_on="geolocation_zip_code_prefix", how="left")
)

main_data["delivery_days"] = (
    main_data["order_delivered_customer_date"] - main_data["order_purchase_timestamp"]
).dt.total_seconds() / 86400.0
main_data["delay_days"] = (
    main_data["order_delivered_customer_date"] - main_data["order_estimated_delivery_date"]
).dt.total_seconds() / 86400.0
main_data["is_late"] = np.where(
    main_data["order_delivered_customer_date"].notna(),
    main_data["delay_days"] > 0,
    np.nan
)

output_csv = os.path.join("dashboard", "main_data.csv")
main_data.to_csv(output_csv, index=False)
print(f"File '{output_csv}' berhasil diekspor! Ukuran: {main_data.shape[0]:,} baris x {main_data.shape[1]} kolom.")
```
